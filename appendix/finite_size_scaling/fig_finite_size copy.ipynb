```python
from load_files_from_cluster import load_cluster_results
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams.update(
    {
        "font.size": 10,
        "text.usetex": True,
        "font.family": "serif",
        "font.serif": ["Times"],
    }
)
import cmasher as csh
aps_figwidth = 6.6
aps_halfwidth = 3.375
```

# 2D

```python
L_2D = [3, 20, 30, 70]
D = 2
kappa_shift = '0'
output_v0_2D = {}
for L_val in L_2D:
    output_v0_2D[L_val] = load_cluster_results(
        f'results_{D}d_v_{kappa_shift}',
        f'results_sigma*_num_reals_100_L{L_val}.h5',
    )
kappa_shift = '3r0'
output_v3_2D = {}
for L_val in L_2D:
    output_v3_2D[L_val] = load_cluster_results(
        f'results_{D}d_v_{kappa_shift}',
        f'results_sigma*_num_reals_100_L{L_val}.h5',
    )
```

```python
z2_v0_2D = {L_val: [np.mean(elem) for elem in output_v0_2D[L_val]['z2']] for L_val in L_2D}
z2_v3_2D = {L_val: [np.mean(elem) for elem in output_v3_2D[L_val]['z2']] for L_val in L_2D}
```

```python

fig, ax = plt.subplots(
    1, 2,
    figsize=(1.2*aps_halfwidth, 0.5*aps_halfwidth),
    sharey=True,
    dpi=350,
    )



cmap = csh.bubblegum
for L_val in L_2D:
    ax[0].plot(
        output_v0_2D[L_val]['sigma_values']*L_val,
        z2_v0_2D[L_val],
        label=f'L={L_val}',
        # marker='o', markersize=2, 
        linewidth=.9,
        color=cmap((L_val - min(L_2D)) / (max(L_2D) - min(L_2D))),
        rasterized=True,
        )
for L_val in L_2D:    
    ax[1].plot(
        output_v3_2D[L_val]['sigma_values']*L_val,
        z2_v3_2D[L_val],
        label=f'L={L_val}',
        # marker='o', markersize=2,
        linewidth=.9,
        color=cmap((L_val - min(L_2D)) / (max(L_2D) - min(L_2D))),
        rasterized=True,
        ) # {L_val**D} sites',

ax[0].set_ylabel(r"$\langle \nu_{\textup{2D}} \rangle$",labelpad=-5)
ax[0].set_xlabel(r'$\sigma (\rho=0)$', labelpad=-11)
ax[1].set_xlabel(r'$\sigma (v=3r_0)$', labelpad=-11)

ax[0].legend(fontsize=8, frameon=False, loc='lower right')
ax[0].set_yticks([-1, 0, 1])

ax[0].set_xlim(0, 0.1)
ax[0].set_ylim(-1.05, 1.05)
ax[1].set_xlim(0, 0.1)

ax[0].set_xticks([0, 0.25, 0.5])
ax[0].set_xticklabels([0, "", r'0.50 $r_0$'])
ax[1].set_xticks([0.25, 0.5])
ax[1].set_xticklabels(["", r'0.50 $r_0$'])
# subplot adjusts:
fig.subplots_adjust(wspace=0.)
```

# 3D

```python
L = [3, 10, 18, 22]
D = 3
kappa_shift = '0'
output_v0 = {}
for L_val in L:
    output_v0[L_val] = load_cluster_results(
        f'results_{D}d_v_{kappa_shift}',
        f'results_sigma*_num_reals_100_L{L_val}.h5',
    )
kappa_shift = '3r0'
output_v3 = {}

for L_val in L:
    output_v3[L_val] = load_cluster_results(
        f'results_{D}d_v_{kappa_shift}',
        f'results_sigma*_num_reals_100_L{L_val}.h5',
    )
```
